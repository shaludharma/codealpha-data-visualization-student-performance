# Student Performance - Data Visualization
Dataset: Students Performance in Exams (1000 students, 8 columns)
Tools: Python, pandas, matplotlib, seaborn

## Setup
Importing libraries and loading the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

## Chart 1: Average Scores by Gender
A bar chart comparing the average math, reading and writing scores of male and female students.

```python
df.groupby("gender")[["math score", "reading score", "writing score"]].mean().plot(kind="bar")
plt.title("Average Scores by Gender")
plt.ylabel("Average Score")
plt.show()
```

## Chart 2: Effect of Test Preparation Course
A bar chart comparing average scores of students who completed the test preparation course with those who did not.

```python
df.groupby("test preparation course")[["math score", "reading score", "writing score"]].mean().plot(kind="bar")
plt.title("Average Scores by Test Preparation Course")
plt.ylabel("Average Score")
plt.show()
```

## Chart 3: Distribution of Math Scores
A histogram showing how math scores are spread across the 1000 students.

```python
sns.histplot(df["math score"], bins=20, kde=True)
plt.title("Distribution of Math Scores")
plt.show()
```

## Chart 4: Math Score by Lunch Type
A box plot comparing math scores of students with standard lunch and free/reduced lunch.

```python
sns.boxplot(x="lunch", y="math score", data=df)
plt.title("Math Score by Lunch Type")
plt.show()
```

## Chart 5: Students by Parental Education Level
A count plot showing how many students belong to each parental education level.

```python
plt.figure(figsize=(10, 5))
sns.countplot(x="parental level of education", data=df)
plt.title("Students by Parental Education Level")
plt.xticks(rotation=30)
plt.show()
```

## Chart 6: Math Score vs Reading Score
A scatter plot showing the relationship between math and reading scores, coloured by gender.

```python
sns.scatterplot(x="math score", y="reading score", hue="gender", data=df)
plt.title("Math vs Reading Score")
plt.show()
```

## Chart 7: Correlation Heatmap
A heatmap showing how strongly math, reading and writing scores are related to each other.

```python
sns.heatmap(df[["math score", "reading score", "writing score"]].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Between Scores")
plt.show()
```

## Conclusion
- Male students scored higher on average in math, while female students scored higher in reading and writing.
- Students who completed the test preparation course scored higher on average in all three subjects.
- Students with standard lunch had a higher median math score (about 69) than students with free/reduced lunch (about 60), and the lowest math scores were mostly from the free/reduced lunch group.
- Most students' parents have "some college", "associate's degree" or "high school" education; "master's degree" is the smallest group.
- Math, reading and writing scores are strongly positively correlated. Reading and writing are the most closely related (0.95), and math has a correlation of about 0.8 with both.
- In the scatter plot, female students tend to score higher in reading than male students with similar math scores.
